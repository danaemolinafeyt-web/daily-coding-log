Drill 2 — add and remove 

Starting from that same list: append a new ticker at the end, insert one at position 0, and insert one in the middle. Then remove one by value, remove one by index, and pop the last one into a variable called dropped — then print a sentence like f"I removed {dropped} from my watchlist."

```python
tickers = ['AAPL', 'QQQM', 'AMD', 'NOW', 'PLTR'] # append new ticker at the end. 
tickers.append('GOLD')
print(tickers)
tickers.insert(0, 'CVLT') # Insert ticker at postion 0
print(tickers)
tickers.insert(3,'ANTHZZX') # insert ticker in the middle
print(tickers)

tickers.remove('PLTR') # removed PLTR from list.
print(tickers)

del tickers[0] # removed ticker by index using del, looked at the book (removes 'CVLT')
print(tickers)

dropped = tickers.pop() #dropped the last item, looked at the book for this. 
print(f"I removed {dropped} from my watchlist.")
```

Drill 3 — sorting 

Make a list of 6 prices (floats). Print them sorted low→high without changing the original, then print the original to prove it's unchanged. Then sort the original permanently, high→low. Then reverse the list and print it. Finally, print the highest price using indexing — no max() allowed.

```python
my_prices = [78.00, 26.00, 33.00, 12.00, 22.00, 279.00]
print("Here is the original list: ")
print(my_prices)

print("Here is the sorted list: ")
print(sorted(my_prices))

my_prices.sort() # sorted lists permanently
print(my_prices)

my_prices.reverse() # list has been reversed. 
print(my_prices)

print(my_prices[0]) # Prints the highest price without using max()
```
