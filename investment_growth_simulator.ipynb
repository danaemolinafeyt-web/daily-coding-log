Step 1: The core function

The compound interest formula:

FV = p(1+r/n)^(nt)

p - principal, what you start with
example: 10000

r - annual interest rate, as a decimal
example: 0.05 for 5%

n - times compounded per year
example: 1 = yearly, 12 = monthly

t - years
example: 3

Warnings:
Rate must be a decimal, 5% is 0.05
Everything in exponents needs parentheses: **(n*t), without parenthese python raises to the power of n first, then multiplys. 

```python
def future_value (p, r, n, t):
    return p*(1+r/n)**(n*t)
future_value(2000, 0.05, 1, 3)
```

The $2000 invested over a period of 3 years, earning yearly at 5% gained $315 dollars. 

Lets now try doing a loop for year- by- year growth. 

```python
p = 2000
r = 0.05
n = 12
years = 10

for t in range(1, years + 1):
    value = future_value(p, r, n, t)
    print(f"Year {t}: {value:,.2f}") 
```

```python
p = 2000
years = 10

#setting up two scenrios

#opt A
r_a, n_a = 0.05, 12

#opt B
r_b, n_b = 0.06, 1

for t in range(1, years +1):
    a = future_value(p, r_a, n_a, t)
    b = future_value(p, r_b, n_b, t)
    diff = b - a
    print(f"Year {t}: A = {a:,.2f} | B = {b:,.2f} | diff = {diff:,.2f}")
```
