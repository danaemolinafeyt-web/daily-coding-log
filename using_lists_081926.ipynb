2b. Using range() to make a List of Numbers
If you want to make a list of numbers, you can convert the results of range directly into a list using the list function. When you wrap list around a call to the range function the output will be a list of numbers.

```python
numbers = list(range(1, 7))
print(numbers)
```

```python
#using range() to skip nummbers

numbers = list(range(3, 17, 3))
print(numbers)
```

4. List Comprehensions
The approach described earlier for generating the list squares consisted of using three or four lines of code a list comprehension allows you to generate the same list and just one line of code

```python
squares = [value**2 for value in range(1, 11)]
print(squares)
```

5. Working with Part of a List
- Slicing a list: 

```python
players = ['charles', 'martina', 'michael', 'florence', 'eli'] 
print(players[0:3]) # from index charles to michael
```

```python
players = ['charles', 'martina', 'michael', 'florence', 'eli'] 
print(players[1:4]) #from index martina to florence
```

```python
players = ['charles', 'martina', 'michael', 'florence', 'eli'] 
print(players[:4]) #without starting index python starts at beginning of list. 
```

```python
players = ['charles', 'martina', 'michael', 'florence', 'eli'] 
print(players[2:]) #this starts at michael and gives the end of the list. 
```

```python
players = ['charles', 'martina', 'michael', 'florence', 'eli'] 
print(players[-3:]) #this outputs the last three players
```

Although you can't modify a table you can assign a new value to a variable that represents a table for example if you wanted to change the dimensions of this rectangle we could redefine the entire couple

1. Loop through and print each day's price as "Day 1: $182.52", "Day 2: $179.80", etc
You'll need the day number and the price together — think about how range() combined with indexing (prices[i]) can get you both at once.
2. Slice out just the most recent 5 days into a new list called recent_week.
3. Print the min, max, and average price of recent_week (average = sum ÷ len)
4. Store the ticker and the max price from recent_week as a tuple called best_day, and print it

```python
# Step 1

prices = [182.52, 179.80, 191.20, 175.10, 188.90, 190.15, 186.40, 193.75, 189.20, 195.60]
for i in range(len(prices)):
    price = prices[i]
    day_number = i + 1
    print(f"Day {day_number}: ${price} ")

# Step 2

recent_week = prices[5:]
print(f"\nthese are the five most recent prices: ")
print(recent_week)

# Step 3

print("\nThis is the Min of Recent Week: ")
print(min(recent_week)) 

print("\nThis is the Max of Recent Week: ")
print(max(recent_week))

average = sum(recent_week)/len(recent_week)

print("\nThis is the Average of Recent Week: ")
print(average)

# Step 4

best_day = ('AAPL', max(recent_week))
print(f"\nThis is the MAX price for the last five days: {best_day[1]}")
```
