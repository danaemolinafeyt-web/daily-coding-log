Summation! for loop in disguise. 

total = 0
FOR i FROM 1 TO n:
    total = total + (something with i)
RETURN total

Σ (i=1 to n) 10·xᵢ reads as: "visit every locker from first to last, multiply what's inside by 10, add it all up."

```python
# Σ(i=1 to 6) 5i

summation = sum(5*i for i in range (1, 7))
print(summation)
```

```python
# lists (the locker pattern)
# Difference: now i is a locker number (an index), not the value itself. You use x[i] to open the locker.

x = [1, 4, 6, 2]
n = len(x)

total = 0
for i in range(0, n):
    total = total + (10 * x[i])
print(total)
```

```python
# B1. x = [2, 5, 9] — compute Σ(i=1 to n) 4·xᵢ

x = [2, 5, 9]
n = len(x)

total = 0
for i in range(0, n):
    total = total + (4 * x[i])
print(total)
```

```python
#  B2. x = [3, 7, 1, 8] — compute Σ(i=1 to n) xᵢ 
# (no multiplier — just add the values)

x = [3, 7, 1, 8]
n = len(x)

total = 0
for i in range(0,n):
    total = total + (x[i])
print(total)
```

```python
# B3. x = [10, 20, 30] — compute Σ(i=1 to n) xᵢ²

x = [10, 20, 30]
n = len(x)

total = 0 
for i in range(0, n):
    total = total + (x[i]**2)
print(total)
```

```python
# SymPy - lowest priotity , enough to recognize pattern. 

from sympy import *
i, n = symbols ('i n')          # Placeholders
summation = Sum(2*i, (i, 1, n)) # Build formula- nothing computed yet. CAPITAL Sum.
up_to_5 = summation.subs(n, 5)  # Plug in a real n - still nothing to compute
print(up_to_5.doit())           # NOW compute
```
